```python
def fantasy_points(player_stats):
    """
    Calculate fantasy points based on player statistics.

    Args:
        player_stats (dict): A dictionary containing player statistics.

    Returns:
        int: The total fantasy points for the player.
    """
    points = 0

    # Example scoring system
    points += player_stats.get('touchdowns', 0) * 6
    points += player_stats.get('yards', 0) // 10
    points += player_stats.get('receptions', 0) * 1
    points -= player_stats.get('turnovers', 0) * 2

    return points
```
